```python
#Ladataan kirjastot
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

#Tarkistetaan työhakemisto, koska suhteellinen polku tulkitaan siitä,
#eikä se aina ole sama kuin tämän notebookin kansio.
print("Työhakemisto:", Path.cwd())

#Etsitään data-tiedosto ensin työhakemiston alta, ja jos sitä ei löydy,
#yritetään löytää se "Harjoitukset/Data"-kansiosta työhakemiston sisältä.
data_path = Path.cwd() / "Data" / "Raw Data.csv"
if not data_path.exists():
    data_path = Path.cwd() / "Harjoitukset" / "Data" / "Raw Data.csv"
if not data_path.exists():
    raise FileNotFoundError(
        f"Datatiedostoa ei löytynyt polusta {data_path}. "
        "Varmista, että työhakemisto on 'Harjoitukset'-kansio tai sen yläkansio."
    )
print("Käytetty datatiedosto:", data_path)

#Luetaan data Pandas DataFrameen
df = pd.read_csv(data_path)
#df['Time (s)'] # Valitsee ajan sarakkeen Time (s)
df

#Piirretään kuvaaja Matplotlib.pyplot -kirjastolla
plt.figure(figsize=(15,8))
plt.plot(df['Time (s)'], df['Linear Acceleration x (m/s^2)'])
plt.plot(df['Time (s)'], df['Linear Acceleration y (m/s^2)'])
plt.plot(df['Time (s)'], df['Linear Acceleration z (m/s^2)'])
plt.plot(df['Time (s)'],df['Absolute acceleration (m/s^2)'])
plt.title('Kiihtyvyysmittaus')
plt.legend(['x', 'y', 'z', 'absoluuttinen'])
plt.grid()
plt.ylabel('Kiihtyvyys (m/$s^2$)')
plt.xlabel('Aika (s)')
plt.show()
#plt.plot(df['Linear Acceleration x (m/s^2)'])

#Koordinaattiakselit:
# x-akseli: ylös - alas
# y-akseli: oikealle - vasemmalle
# z-akseli: eteen - taakse




```
